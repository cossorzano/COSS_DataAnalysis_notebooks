The objective of this exercise is to solve an optimization problem.
Whis is the following, a farmer has 100 hectares divided between wheat and corn. The objective is to MAXIMIZE the profits.
Variables:
- $x_1$ = wheat
- $x_2$ = corn
- $y$ = total benefit

Constrains:
- land: $x_1+x_2 \leq 100$
- budget: $120x_1 + 200x_2 \leq 20000$
- work: $10x_1 + 20x_2 \leq 1200$
- no negative: $x_1\geq 0, x_2\geq 0$

Objective Function:
- $\max y= 300x_1+400x_2$

Firslty we import the libraries needed for the solution of the problem. In this case we are using pulp, which allows users to define variables, constraints, and objective functions, and then solve these problems using various solvers.

```python
!pip install pulp
```

```python
import pulp as plp
```

Now we set our problem and variables. Indicating that we want to maximize a problem.

```python
model_resource = plp.LpProblem("Resources", plp.LpMaximize)
x1 = plp.LpVariable("trigo", 0, )
x2 = plp.LpVariable("maiz", 0)
```

Next we set out objective function, all toghether with the constrains. The main difference between the objective function and the constrains is that the constrains have a more/less or equal sign. We also solve this problem.

```python
model_resource += 300*x1+400*x2, "Objective function"
model_resource += x1+x2 <= 100
model_resource += 120*x1 + 200*x2 <= 20000
model_resource += 10*x1 + 20*x2 <= 1200

model_resource.solve()
```

We visualize the results

```python
print("Estado de la solución:",plp.LpStatus[model_resource.status])

print(f"Cantidad óptima de trigo: {x1.varValue}")
print(f"Cantidad óptima de maiz: {x2.varValue}")

print(f"Beneficio total máximo: ${plp.value(model_resource.objective)}")
```

**Second part**

For this second part we wanted to find the parameter where the solution stays optimal. In this code we search for the range of values for a parameter in the objective function that allows the optimal solution of the linear programming problem to remain the same, which is
$x_1=80$ and $x_2=20$. To do this we define a model with an objective function and constraints that limit the possible combinations of $x_1$ and $x_2$. Then we perform two searches, the first varies the coefficient multiplying $x_1$ from 200 to 600 in steps of 10 and for each value we solve the problem checking if the optimal solution stays the same. If it does we save that value in a list. The second search varies the coefficient multiplying $x_2$ following the same procedure. Finally we print the lists with the values that keep the optimal solution and the status of the problem.

```python
lista_rango_x = []
lista_rango_y = []

for x in range(200,610,10):
    model_resource = plp.LpProblem("Resources", plp.LpMaximize)
    model_resource += x*x1+400*x2
    model_resource += x1+x2 <= 100
    model_resource += 120*x1 + 200*x2 <= 20000
    model_resource += 10*x1 + 20*x2 <= 1200
    model_resource.solve(plp.PULP_CBC_CMD(msg=False))
    if x1.varValue == 80 and x2.varValue == 20:
        lista_rango_x.append(x)
    else:
        next

print("List of values that continue optimizing the problem:",lista_rango_x)

for y in range(200,610,10):
    model_resource = plp.LpProblem("Resources", plp.LpMaximize)
    model_resource += 300*x1+y*x2
    model_resource += x1+x2 <= 100
    model_resource += 120*x1 + 200*x2 <= 20000
    model_resource += 10*x1 + 20*x2 <= 1200
    model_resource.solve(plp.PULP_CBC_CMD(msg=False))
    if x1.varValue == 80 and x2.varValue == 20:
        lista_rango_y.append(y)
    else:
        next

print("List of values that continue optimizing the problem:",lista_rango_y)
print("Estado de la solución:",plp.LpStatus[model_resource.status])
```
