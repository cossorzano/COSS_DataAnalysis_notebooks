In this notebook I will resolve a *Linear Programming* problem. *Linear Optimization* is where both the objective function and contraints are linear. These problems can be efficiently dolved using techniques such as the simplex method or interior-point methods.

*Linear Programming* consists of :
- **Decision variables** are the variables $x$ and $y$, which decide the output of the linear programming problem and represent the final solution.
- **Objective function**, generally respresented by $Z$, is the linear function that needs to be optimized according to the given condition to get the final solution.
- **Constraints**
- **Non-negative restrictions** ($x\geq 0$, $y\geq 0$)

**<h4>Exercise: Logistics</h4>**

A company must ship products from two warehouses to three retail stores. The goal is to minimize transportation cost while fulfilling store demand and not exceeding warehouse supply.

Le $x_{ij}$ be the number of units shipped from warehouse $i$ to store $j$:
- $x_{11}$, $x_{12}$, $x_{13}$ : from warehouse 1 to stores 1, 2, 3
- $x_{21}$, $x_{22}$, $x_{23}$ : from warehouse 2 to stores 1, 2, 3

| From \ To   | Store 1 | Store 2 | Store 3 | Suply |
|-------------|---------|---------|---------|-------|
| Warehouse 1 | €4      | €6      | €9      | 100   |
| Warehouse 2 | €5      | €4      | €7      | 120   |
| Demand      | 80      | 70      | 50      |



Objective : Minimize total shipping cost.

**<u>Constraints</u>**

- _Supply constraints_:

    $x_{11} + x_{12} + x_{13} \leq 100$   (Warehouse 1)

    $x_{21} + x_{22} + x_{23} \leq 120$   (Warehouse 2)

- _Demand contraints_:

    $x_{11} + x_{21} = 80$ (Store 1)

    $x_{12} + x_{22} = 70$ (Store 2)

    $x_{13} + x_{23} = 50$ (store 3)

- _Non-negativity_:

    $x_{ij} \geq 0$ for all $i, j$



<u>Objective function:</u>

Minimize cost $= 4x_{11} + 6x_{12} + 9x_{13} + 5x_{21} + 4x_{22} + 7x_{23}$

First of all, we can see that the sum of supply $= 100 + 120 = 220 >$ sum of demand $= 220$, which means there is excess capacity.

---

**<u>We assign by minimum cost</u>**

*Order of costs (from lowest to highest):*

`4 (W1-S1 and W2-S2), 5 (W2-S1), 6 (W1-S2), 7 (W2-S3), 9 (W1-S3)`

- *Store 1 (demand = 80):*
    - W1->S1 (4€/u)
    - W2->S1 (5€/u)

    In the end we take the "W1->S1" since it has a minimum cost in store 1, so $x_{11} = 80$.

- *Store 2 (demand = 70):*
    - W1->S2 (4€/u)
    - W2->S2 (6€/u)

    In the end we take "W1->S2" since it is the minimum cost in store 2, so $x_{22} = 70$.

- *Store 3 (demand = 50):*
    - W1->S3 (9€/u)
    - W2->S3 (7€/u)

    In the end we take "W2->S3" since it is the minimum cost in store 3, so $x_{23} = 50$.


- $x_{12} = x_{13} = x_{21} = 0$

*Constraints:*
- W1 : $x_{11} + x_{12} + x_{13} = 80 \leq 100$
- W2 : $0 + 70 + 50 = 120 \leq 120$
- Demand : $S1=80, S2=70, S3=50$

*<u>We solve the total cost of the solution:</u>*

$C = 4*80 + 6*0 + 9*0 + 5*0 + 4*70 + 7*50 = 950$€

After having calculated the total cost of the solution, <u>we calculate the values of the x's</u>



```python
from pulp import * # We import the pulp library

# Create the minimization problem
prob = LpProblem("Logistic_Problem", LpMinimize)
# With LpProblem() we create a new optimization problem
    # "Logistics_Problem" would be the name we give to the problem
    # LpMinimize to indicate that we want to minimize it

# Define the decision variables
# With LpVariable(name of the variable, lowbound = 0) we define the decision variables, lowbound = 0 means that the variable cannot be negative
x11 = LpVariable("x11", lowBound=0)  # From warehouse 1 to store 1
x12 = LpVariable("x12", lowBound=0)  # From warehouse 1 to store 2
x13 = LpVariable("x13", lowBound=0)  # From warehouse 1 to store 3
x21 = LpVariable("x21", lowBound=0)  # From warehouse 2 to store 1
x22 = LpVariable("x22", lowBound=0)  # From warehouse 2 to store 2
x23 = LpVariable("x23", lowBound=0)  # From warehouse 2 to store 3
```

```python
# Objective function: Minimize total cost
prob += 4*x11 + 6*x12 + 9*x13 + 5*x21 + 4*x22 + 7*x23, "Total_cost"
# We add the objective function to the problem, and "Total_Cost" is the name we give to the objective function.
# With this, we want to find the values of x_11, x_12, x_13, x_21, x_22, and x_23 so that the cost is minimal.

# We add the constraints to the problem:
# Supply restrictions (warehouses)
prob += x11 + x12 + x13 <= 100, "Warehouse_Offer_1"
prob += x21 + x22 + x23 <= 120, "Warehouse_Offer_2"

# Demand restrictions (stores)
prob += x11 + x21 == 80, "Demand_Store_1"
prob += x12 + x22 == 70, "Demand_Store_2"
prob += x13 + x23 == 50, "Demand_Store_3"
```

```python
# Solve the problem
prob.solve()

print("State:", LpStatus[prob.status]) # As we can see it is optimal
print("Minimum total cost:", value(prob.objective)) # And with this we can see the minimum cost

# Variable values
for v in prob.variables():
    print(v.name, "=", v.varValue)

print("\nSlacks in the restrictions:")
for name, constraint in prob.constraints.items():
    print(f"{name}: slack = {constraint.slack}, shadow price (pi) = {constraint.pi}")
```

- **State of the model: Optimal**

    Indicates that the model found an optimal solution that minimizes total cost while meeting all supply and demand constraints.
    That is, there is no other shipping combination with a lower cost.

- **Minimum total cost: 950€**

    This is the lowest possible total transportation cost to meet the entire demand of the three stores using the two available warehouses.
    It corresponds to the optimal value of the objective function:
  $$C=4x_{11}+6x_{12}+9x_{13}+5x_{21}+4x_{22}+7x_{23}=950$$

- **Slacks in the restrictions**

    The _slack_ value indicates how much “free space” remains in a constraint. The _shadow price_ shows how much the total cost would change if the constraint limit were changed by one unit.

    In the results, Warehouse_Offer_1 has a slack of 20 and a shadow value of 0, so its capacity does not affect the cost. The other supply and demand constraints; demands have positive shadow prices (4 or 7), indicating that increasing demand would increase the cost, with store 3 being the most expensive to supply.

- **Decision variables**

    In the optimal solution to the logistics model, the decision variables show that only three transportation routes are used. The shipment from *warehouse 1 to store 1* $(x_{11} = 80)$ completely satisfies this store's demand at the lowest possible cost (4€/unit). Similarly, *store 2 receives its 70 units from warehouse 2* $(x_{22} = 70)$, also at a minimum cost of 4€/unit, and *store 3 is supplied entirely by warehouse 2* $(x_{23} = 50)$, whose unit cost is 7€. The other possible routes $(x_{12}, x_{13},$ and $x_{21})$ are not used in the final solution, as they would increase the total transportation cost. Overall, this allocation ensures that all demand is covered at the lowest overall cost of 950€.

    The model selects only the cheapest possible routes:
    - Warehouse 1 --> Store 1 (4€/u)
    - Warehouse 2 --> Store 2 (4€/u)
    - Warehouse 2 --> Store 3 (7€/u)


**_Conclusion_**

To sum up, the linear programming model allowed us to determine an optimal transportation allocation that minimizes total cost to €950, meeting all supply and demand constraints. The solution shows that store 1 is supplied entirely from warehouse 1, while stores 2 and 3 receive their products from warehouse 2, using only the most economical routes. This distribution guarantees total supply at the lowest possible cost, avoiding unnecessary shipments via more expensive routes. In short, the model achieves efficient management of logistics resources, ensuring a balance between costs and demand fulfillment.
