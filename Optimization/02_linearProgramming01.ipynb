```python
import numpy as np
import matplotlib.pyplot as plt

# Define the grid of values for x1 (wheat) and x2 (corn)
x1 = np.linspace(0, 100, 500)
x2 = np.linspace(0, 100, 500)
X1, X2 = np.meshgrid(x1, x2)

# Define constraints
land = X1 + X2 <= 100                             # Land constraint
budget = 120 * X1 + 200 * X2 <= 20000            # Budget constraint
labor = 10 * X1 + 20 * X2 <= 1200                # Labor constraint

# Feasible region: intersection of all constraints
feasible = land & budget & labor

# Plot
plt.figure(figsize=(8, 8))
plt.contourf(X1, X2, feasible, levels=[0.5, 1], colors=["#3399ff"], alpha=0.7)

# Constraint boundary lines
plt.plot(x1, 100 - x1, label='Land: $x_1 + x_2 = 100$')
plt.plot(x1, (20000 - 120 * x1) / 200, label='Budget: $120x_1 + 200x_2 = 20000$')
plt.plot(x1, (1200 - 10 * x1) / 20, label='Labor: $10x_1 + 20x_2 = 1200$')

# Profit lines: y = 300x1 + 400x2 = k
for k in [20000, 25000, 30000]:
    plt.plot(x1, (k - 300 * x1) / 400, '--', color='gray', label=f'Profit: {k} €')

# Plot formatting
plt.xlim(0, 100)
plt.ylim(0, 100)
plt.xlabel('$x_1$ (Wheat area in ha)')
plt.ylabel('$x_2$ (Corn area in ha)')
plt.title('Feasible Region and Profit Lines for Resource Allocation')
plt.legend()
plt.grid(True)
plt.show()
```
