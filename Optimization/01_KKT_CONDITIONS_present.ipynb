The KKT conditions are necessary conditions for a solution in nonlinear programming to be optimal, given some regularity assumptions.
They extend the method of Lagrange multipliers to problems with inequality constraints.
Moreover, if the problem is convex, these conditions are also sufficient for global optimality.

The KKT conditions consist of four parts:
- Stationarity: The gradient of the Lagrangian must vanish at the optimum.
- Primal feasibility: The constraints of the original problem must be satisfied.
- Dual feasibility: The Lagrange multipliers must be non-negative.
- Complementary slackness: Each multiplier times its corresponding constraint must equal zero

### **EXAMPLE: Minimize a function with inequality constraint** ###

### **Importing libraries** ###
We start by importing the Sympy library. It will allow us to work with symbolic variables and equations.

```python
import sympy as sp
import numpy as np
```

### **Defining variables, functions, and Lagrangian** ###
Here we define:
- The decision variables: $x_1, x_2$.
- The objective function: $f(x_1, x_2) = x_1^2 + x_2^2$.
- The restriction: $g(x_1, x_2) = 1 - x_1 - x_2 ≤ 0$.
- The Lagrangian: $L = f + λg$.
- The Lagrange multiplier: $λ$.

```python
x1, x2, lamb = sp.symbols('x1 x2 lambda')

f = x1**2 + x2**2
g = 1 - x1 - x2
L = f + lamb * g
```

### **4 Conditions** ###
Now we apply the four KKT conditions:
- Stationarity: we compute the partial derivatives of $L$ with respect to $x_1$ and $x_2$, set them equal to zero and solve for the variables.
- Primal feasibility: check that the constraint $g(x1, x2) ≤ 0$ is satisfied.
- Dual feasibility: impose the condition $λ ≥ 0$.
- Complementary slackness: impose the condition $λg(x_1, x_2) = 0$

```python
#1. Stationarity
dL_dx1 = sp.diff(L, x1)
dL_dx2 = sp.diff(L, x2)

stat_eq1 = sp.Eq(dL_dx1, 0)
stat_eq2 = sp.Eq(dL_dx2, 0)

sol_x1 = sp.solve(stat_eq1, x1)[0]
sol_x2 = sp.solve(stat_eq2, x2)[0]

print("From stationarity:")
print(f"x1 = {sol_x1}")
print(f"x2 = {sol_x2}")

#2. Primal feasibility
primal_feas = g <= 0
print("\nPrimal feasibility:")
print(primal_feas)

#3. Dual feasibility
dual_feas = lamb >= 0
print("\nDual feasibility:")
print(dual_feas)

#4. Complementary slackness
compl_slackness = sp.Eq(lamb * g, 0)
print("\nComplementary slackness:")
print(compl_slackness)
```

### **Lambda** ###
Next, we solve the complementary slackness condition to find possible values of $λ$.
From these solutions, we select the value that satisfies dual feasibility ($λ ≥ 0$) and primal feasibility ($g ≤ 0$).
This gives us the correct value of the multiplier, which is 1, and which we then substitute again into the solutions for $x_1$ and $x_2$.

```python
#Now we have to look for the value of lambda
g_sub = g.subs({x1: sol_x1, x2: sol_x2})
lambda_sols = sp.solve(sp.Eq(lamb * g_sub, 0), lamb)
print(f"\nLambda solustions = {lambda_sols}") #We have 0 and 1, but we only want the value 1.

lambda_val = np.NaN
for sol in lambda_sols:
    if sol >=0 and g_sub.subs(lamb, sol) <= 0:
        lambda_val = sol
print(f"Final lambda value = {lambda_val}") #Now, lambda = 1.

x1_opt = sol_x1.subs(lamb, lambda_val)
x2_opt = sol_x2.subs(lamb, lambda_val)

print(f"\nSet lambda = {lambda_val}")
print(f"Optimal x1 = {x1_opt}")
print(f"Optimal x2 = {x2_opt}")
```

### **Solutions** ###
Finally, we compute:
- The optimal point ($x_1^*, x_2^*$).
- The minimum value of the objective function at this point.

```python
#Finally, these are the final solutions.
print(f"\nSolutions:")
print(f"Optimal point: ({x1_opt}, {x2_opt})")
min_value = f.subs({x1: x1_opt, x2: x2_opt})
print(f"Minimum value f(x1, x2) = {min_value}")
```
